```python
# sort 'ObservingMethod-migration.csv' by 'disposition'
import csv
from pathlib import Path

CSV_PATH = Path('ObservingMethod-migration.csv')
if not CSV_PATH.exists():
    raise FileNotFoundError(f'File not found: {CSV_PATH}')

with CSV_PATH.open(newline='', encoding='utf-8') as fh:
    reader = csv.DictReader(fh)
    fieldnames = reader.fieldnames
    rows = list(reader)

# Sort rows by the 'disposition' column (empty values first)
rows.sort(key=lambda r: (r.get('disposition') or ''))

with CSV_PATH.open('w', newline='', encoding='utf-8') as fh:
    writer = csv.DictWriter(fh, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(rows)

print('Sorted and saved', CSV_PATH)
```

```python
# sort 'ObservingMethod-proposed.csv' by 'reviewStatus'
"""Sort ObservingMethod-proposed.csv rows by the 'reviewStatus' column in-place."""
import csv
from pathlib import Path

CSV_PATH = Path("ObservingMethod-proposed.csv")

def main():
    if not CSV_PATH.exists():
        raise SystemExit(f"File not found: {CSV_PATH}")

    with CSV_PATH.open(newline='', encoding='utf-8') as fh:
        reader = csv.DictReader(fh)
        fieldnames = reader.fieldnames
        rows = list(reader)

    # Sort by reviewStatus (empty values sort first)
    rows.sort(key=lambda r: (r.get('reviewStatus') or ''))

    with CSV_PATH.open('w', newline='', encoding='utf-8') as fh:
        writer = csv.DictWriter(fh, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(rows)

    print(f"Sorted {CSV_PATH.name} by 'reviewStatus' and saved.")

if __name__ == '__main__':
    main()
```
